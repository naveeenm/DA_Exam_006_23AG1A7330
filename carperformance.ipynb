```
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from google.colab import files


data = pd.read_csv('car_price_performance.csv')
```

```
brand_avg_price = data.groupby('Brand')['Price'].mean().sort_values()
brand_to_encoded = {brand: idx for idx, brand in enumerate(brand_avg_price.index)}
data['Encoded_Brand'] = data['Brand'].map(brand_to_encoded)
print("After Label Encoding (Task 1):")
print(data[['Brand', 'Price', 'Horsepower', 'Fuel Efficiency', 'Encoded_Brand']].head())
```

```
scaler = MinMaxScaler()
data['Scaled_Horsepower'] = scaler.fit_transform(data[['Horsepower']])
data['Scaled_Fuel_Efficiency'] = scaler.fit_transform(data[['Fuel Efficiency']])
data['Scaled_Brand'] = scaler.fit_transform(data[['Encoded_Brand']])
print("After Feature Scaling (Task 2):")
print(data[['Brand', 'Price', 'Horsepower', 'Fuel Efficiency', 'Encoded_Brand',
            'Scaled_Horsepower', 'Scaled_Fuel_Efficiency', 'Scaled_Brand']].head())
```

```
data['Performance_Index'] = (data['Horsepower'] * 0.7 + data['Fuel Efficiency'] * 0.3) / data['Price']
print("After Performance Index Calculation (Task 3):")
print(data[['Brand', 'Price', 'Horsepower', 'Fuel Efficiency', 'Encoded_Brand',
            'Scaled_Horsepower', 'Scaled_Fuel_Efficiency', 'Scaled_Brand',
            'Performance_Index']].head())
```

```
# Task 4: Create final table and save to CSV
final_columns = [
    'Brand', 'Price', 'Horsepower', 'Fuel Efficiency',
    'Encoded_Brand', 'Scaled_Horsepower', 'Scaled_Fuel_Efficiency',
    'Scaled_Brand', 'Performance_Index'
]

final_data = data[final_columns]
print("Final Table Preview:")
print(final_data.head())
final_data.to_csv('scaled_car_data.csv', index=False)
print("\nScaled data saved to 'scaled_car_data.csv'. Downloading file...")
files.download('scaled_car_data.csv')
```
